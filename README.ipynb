# Topological Analysis and Vulnerability Identification in Software Supply Chains

**Case Study:** Flowise (AI Agent Orchestration Framework)

## Project Overview
This project maps and analyzes the software supply chain of the `flowise` ecosystem (v3.1.2) using advanced Network Science methodologies. By modeling package dependencies as a **Directed Acyclic Graph (DAG)**, we identify structural vulnerabilities, critical bottlenecks, and single points of failure that could be exploited in targeted cyber attacks (e.g., zero-day exploits, malicious package injections).

## Project Phases Completed Successfully:

1. **Environment & Data Ingestion (Phases 1-2)**
   - Extracted the complete transitive dependency graph using the Google Open Source Insights API (`deps.dev`).
   - Constructed a NetworkX **Directed Graph** (`DiGraph`) compromising ~2,171 nodes and >5,400 directed edges.

2. **Topological Vulnerability Assessment (Phase 3)**
   - Calculated **Betweenness Centrality** to identify "bridge" packages. These structural bottlenecks are highly critical, as their compromise maximizes the "blast radius" of a supply chain attack.

3. **Ecosystem Clustering & Functional Segmentation (Phases 4-5)**
   - Displayed the blast radius (Ego-Network) for individual critical vulnerabilities.
   - Applied the **Clauset-Newman-Moore greedy modularity maximization** algorithm to detect distinct functional communities within the chaos of the dependency graph.

4. **Micro-Level Edge Betweenness - Girvan-Newman Approach (Phase 5b)**
   - Calculated global **Edge Betweenness Centrality** to find the specific dependency "cables" (edges) that carry the most structural load between clusters.

5. **Scale-Free Architecture Modeling (Phase 6)**
   - Analyzed the In-Degree distribution using Log-Log scaling, confirming a Power Law distribution ($P(k) \sim k^{-\gamma}$). This verifies the network is robust to random failures but extremely fragile to targeted hub attacks.

6. **Targeted Disruption Simulation (Phase 7)**
   - Quantified total resilience by sequentially removing high-centrality hubs and monitoring the degradation of the Largest Weakly Connected Component (LWCC).

7. **Empirical Validation / Threat Intel (Phase 8)**
   - Cross-referenced our theoretical topological bottlenecks with active, real-world vulnerabilities (CVEs) from GitHub Advisories. This successfully verified that high-centrality components actively correspond to Critical (CVSS 9.0+) level risk in the real world.

## Methodological Justification: Algorithm Selection & Rationale

Why did we choose these specific methodologies, and how do they benefit the research compared to alternatives?

### 1. Why a Directed Graph (`nx.DiGraph`) over an Undirected Graph?
* **The Choice:** We modeled the ecosystem exclusively as a Directed Graph.
* **The Benefit:** In a software supply chain, vulnerability flow is strictly one-way. If package A depends on B, a zero-day in B compromises A, but a vulnerability in A does *not* compromise B. 
* **Why not the alternative?** Using an undirected graph would create false bidirectional paths (false positives), leading the algorithms to point out incorrect attack vectors and calculate an inaccurate blast radius.

### 2. Why Betweenness Centrality over Degree Centrality?
* **The Choice:** We prioritized Betweenness Centrality to map out structural vulnerabilities.
* **The Benefit:** It tracks the shortest paths across the network, accurately identifying "bridge" nodes (Bottlenecks) that connect entirely different developer communities. Taking down a bridge node maximizes the "blast radius" of a cyberattack across diverse functional domains.
* **Why not the alternative?** Relying purely on **In-Degree Centrality** (who has the most dependents) is a trap. A package can have hundreds of dependents but only exist within a single, isolated community. A package with a lower degree but a high Betweenness measure is a hidden **Single Point of Failure (SPOF)** for the broader ecosystem.

### 3. Why the Fast Greedy Modularity Algorithm over classic Girvan-Newman for Community Detection?
* **The Choice:** Clauset-Newman-Moore greedy modularity maximization ($O(N \log^2 N)$).
* **The Benefit:** This enabled us to efficiently and accurately split the 2,171-node, 5,400-edge chaotic "hairball" graph into meaningful functional clusters (e.g., networking, DB utilities) in seconds rather than hours.
* **Why not the alternative?** The traditional full **Girvan-Newman algorithm** (which relies on continuous edge removal and recalculation with $O(E^2 N)$ time complexity) is computationally prohibitive and unscalable for large software supply chains. However, we did strategically extract its core mechanism—**Edge Betweenness** (Phase 5b)—to analyze the bridging dependencies between our most vulnerable clusters.
